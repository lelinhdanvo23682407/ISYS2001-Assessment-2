# Household Fair-Share Budget Assistant

## Step 1: Understand the Problem

**Problem:** People sharing a household often divide common expenses equally, regardless of differences in their incomes. This may place a greater financial burden on the person earning less. Although equal splitting works for some households, others may prefer an income-proportional approach to managing shared expenses.

**Target Users:** Two individuals living together who want to divide their shared household expenses according to their monthly incomes.

**Purpose:** The application aims to help users calculate how much each person should contribute to shared household bills based on their income proportions rather than automatically splitting every expense equally.

**Proposed Solution:** Users will enter their names, monthly take-home incomes, and shared household expenses, such as rent, electricity, water, and internet. Python will calculate each person's income percentage and allocate the expenses accordingly. A Gemini-powered finance assistant will explain the calculated results and answer relevant household budgeting questions without replacing the Python calculations.

**Expected Benefit:** The application provides a transparent and consistent method for allocating shared household expenses, helping users understand their financial contributions and make informed decisions about shared financial responsibilities.

## Step 2: Identify Inputs and Outputs

### Inputs

The Household Fair-Share Budget Assistant requires the following inputs:

- **Person A:** Name and monthly take-home income (AUD).
- **Person B:** Name and monthly take-home income (AUD).
- **Shared Household Expenses:** Expense names and monthly amounts stored in a dictionary.
- **Optional AI Question (Planned):** A question related to the household expense-sharing results or method.

During the initial implementation, predefined example values will be passed into Python functions. Interactive user inputs will be introduced when the Gradio interface is implemented.

### Outputs

The Python calculator produces:

- Total combined monthly household income.
- Each person's income contribution percentage.
- Total monthly shared household expenses.
- Each person's contribution towards every shared expense.
- Total monthly contribution required from each person.
- Error messages for invalid inputs.
- A dictionary containing the calculated results for reuse.

An AI-generated explanation will be added when Gemini integration is implemented.

Monetary results are displayed in Australian dollars (AUD) with two decimal places.

### Processing Rules

**1. Validate Inputs**

- Check that neither person's name is an empty string.
- Convert monetary inputs into numeric values using `float()`.
- Reject negative amounts, non-finite values, amounts above AUD 1 billion, and values containing fractions of a cent.
- Convert accepted monetary amounts into integer cents using `int()` and arithmetic.
- Use `try/except` and `raise ValueError` to handle invalid monetary inputs.

The current validation is intended for ordinary numeric and string inputs. It does not handle every possible Python data type or floating-point precision edge case. Names containing only spaces are not currently rejected.

**2. Calculate Combined Income**

Combined Income = Person A's Income + Person B's Income

The combined monthly income must be greater than zero to avoid division by zero.

**3. Calculate Income Proportions**

Person A's Proportion = Person A's Income / Combined Income

Person B's Proportion = Person B's Income / Combined Income

If one person has zero income while the other has a positive income, the contribution proportions are 0% and 100%, respectively.

**4. Allocate Shared Expenses**

- Store expenses in a dictionary using expense names as keys and amounts as values.
- Use a `for` loop to process each expense.
- Convert each expense amount into integer cents.
- Calculate Person A's proportional contribution using integer arithmetic.
- Round Person A's contribution to the nearest whole cent using floor division (`//`), rounding exact half-cent cases up.
- Calculate Person B's contribution as the remaining cents.
- Store the allocations in a nested dictionary.

This approach ensures that both contributions add up to the original expense amount in cents.

**5. Calculate and Verify Totals**

- Add the original expense amounts to calculate total shared expenses.
- Add Person A's and Person B's allocated contributions separately.
- Verify that both contributions add up to the total shared expenses.
- If no expenses are provided, return zero expenses and zero contributions.

**6. Display Results**

Display:

- Each person's income contribution percentage.
- The breakdown of contributions for every expense.
- Combined household income.
- Total monthly shared expenses.
- Each person's total monthly contribution.

Use f-strings to display monetary amounts with two decimal places.

**7. Planned AI Integration**

Gemini will receive the Python-calculated results and an optional user question.

It will explain the allocation method without performing or modifying the financial calculations.

If Gemini is unavailable, the Python calculator will continue to provide its results.

### Assumptions and Limitations

- The application supports exactly two individuals.
- Both individuals agree to use income-proportional expense sharing.
- All income and expense values represent monthly amounts in AUD.
- Income refers to take-home income rather than gross income.
- Monetary amounts must be between AUD 0 and AUD 1 billion.
- Monetary inputs must not contain fractions of a cent.
- Names containing only spaces are not currently rejected.
- Expense names are used as dictionary keys. Identical duplicate keys may be overwritten before validation.
- The program uses `float()` to interpret monetary inputs before converting them into integer cents, which introduces some floating-point precision limitations.
- Person A receives the additional cent when an exact half-cent allocation occurs.
- The application does not consider individual debts, savings goals, or other personal financial obligations.
- Interactive inputs and Gemini explanations are planned features, not yet part of the core calculator.
- AI-generated explanations will be informational and will not replace professional financial advice.

## Step 3: Work an Example by Hand

### Example Scenario

Two individuals share a household and want to divide their monthly expenses according to their respective monthly take-home incomes rather than splitting every expense equally.

The following example demonstrates the expected calculations manually before implementing them in Python.

### 1. Monthly Income

| Person | Monthly Income (AUD) |
|---|---:|
| Person A | 3,000.00 |
| Person B | 2,000.00 |
| **Combined Income** | **5,000.00** |

Combined Income = 3,000 + 2,000 = **AUD 5,000.00**

### 2. Shared Monthly Expenses

| Expense | Amount (AUD) |
|---|---:|
| Rent | 1,200.00 |
| Electricity | 200.00 |
| Internet | 100.00 |
| **Total Expenses** | **1,500.00** |

### 3. Calculate Income Proportions

**Person A:**

3,000 / 5,000 = 0.60 = **60%**

**Person B:**

2,000 / 5,000 = 0.40 = **40%**

Therefore, Person A contributes 60% and Person B contributes 40% towards each shared expense.

### 4. Convert Monetary Values into Integer Cents

The application converts monetary values into integer cents to avoid fractional-cent payments.

- Person A's income: 3,000 × 100 = 300,000 cents.
- Person B's income: 2,000 × 100 = 200,000 cents.
- Combined income: 500,000 cents.
- Rent: 120,000 cents.
- Electricity: 20,000 cents.
- Internet: 10,000 cents.

### 5. Allocate Each Expense

The allocation follows these rules:

**Person A's Contribution (cents):**

(2 × Expense Cents × Person A's Income Cents + Combined Income Cents) // (2 × Combined Income Cents)

The `//` operator performs floor division. This formula rounds Person A's proportional contribution to the nearest whole cent, with exact half-cent cases rounded up.

**Person B's Contribution (cents):**

Expense Cents − Person A's Contribution Cents

**Rent**

Person A: 120,000 × 300,000 / 500,000 = 72,000 cents = AUD 720.00

Person B: 120,000 − 72,000 = 48,000 cents = AUD 480.00

**Electricity**

Person A: 20,000 × 300,000 / 500,000 = 12,000 cents = AUD 120.00

Person B: 20,000 − 12,000 = 8,000 cents = AUD 80.00

**Internet**

Person A: 10,000 × 300,000 / 500,000 = 6,000 cents = AUD 60.00

Person B: 10,000 − 6,000 = 4,000 cents = AUD 40.00

### 6. Expense Allocation Summary

| Expense | Person A (60%) | Person B (40%) |
|---|---:|---:|
| Rent | AUD 720.00 | AUD 480.00 |
| Electricity | AUD 120.00 | AUD 80.00 |
| Internet | AUD 60.00 | AUD 40.00 |
| **Total Contribution** | **AUD 900.00** | **AUD 600.00** |

### 7. Verify the Results

**Income proportions:**

60% + 40% = 100%

**Total shared expenses:**

AUD 1,200.00 + AUD 200.00 + AUD 100.00 = AUD 1,500.00

**Total allocated contributions:**

AUD 900.00 + AUD 600.00 = AUD 1,500.00

The total contributions match the total shared household expenses.

### 8. Rounding Example

Consider an expense of AUD 100.01 and equal monthly incomes of AUD 1,000 each.

The expense is converted into 10,001 cents.

Both individuals have a 50% income proportion.

Person A's unrounded allocation:

10,001 × 0.50 = 5,000.5 cents.

The program rounds Person A's contribution up to 5,001 cents.

Person B receives the remaining amount:

10,001 − 5,001 = 5,000 cents.

**Final allocation:**

- Person A: AUD 50.01.
- Person B: AUD 50.00.
- Total: AUD 100.01.

This ensures that both contributions match the original expense amount.

## Step 5: Implement the Core Python Functions

### Implementation Approach

The Household Fair-Share Budget Assistant uses Python concepts practised in ISYS2001, including variables, arithmetic operators, conditional statements, loops, functions, dictionaries, and error handling.

The application separates input validation, income calculations, expense allocation, total verification, and result presentation into reusable functions.

Monetary amounts are converted into integer cents for calculation. The application allocates each expense according to the income proportions, rounds Person A's contribution to the nearest cent using integer arithmetic, and assigns the remaining cents to Person B.

The core calculations will be tested before integrating the user interface and Gemini API.

```python
def validate_name(name):
    # Check that the name is not empty.
    if name == "":
        raise ValueError("Name cannot be empty.")

    # Return the valid name.
    return name


def to_cents(amount):
    # Convert the input to a number.
    # try/except prevents invalid text from crashing the program.
    try:
        amount = float(amount)
    except ValueError:
        raise ValueError("Please enter a valid number.")

    # Reject negative, non-finite, or excessively large values.
    # The upper limit also keeps calculations within the app's scope.
    if not (amount >= 0 and amount <= 1000000000):
        raise ValueError("Amount must be between 0 and 1 billion AUD.")

    # Convert AUD into whole cents.
    # Adding 0.5 rounds positive values to the nearest integer.
    cents = int(amount * 100 + 0.5)

    # Reject amounts containing fractions of a cent.
    # A small tolerance accounts for floating-point representation.
    if abs(amount * 100 - cents) > 0.000001:
        raise ValueError("Enter no more than two decimal places.")

    return cents
```

```python
def calculate_proportions(income_a, income_b):
    # Convert both incomes into cents.
    income_a_cents = to_cents(income_a)
    income_b_cents = to_cents(income_b)

    # Add the two incomes.
    combined_income = income_a_cents + income_b_cents

    # Prevent division by zero.
    if combined_income == 0:
        raise ValueError("Combined income must be greater than zero.")

    # Calculate each person's proportion of total income.
    proportion_a = income_a_cents / combined_income
    proportion_b = income_b_cents / combined_income

    # Return both proportions.
    return proportion_a, proportion_b
```

```python
def allocate_expenses(expenses, income_a, income_b):
    # Create a dictionary to store expense allocations.
    allocations = {}

    # Convert incomes into integer cents.
    income_a_cents = to_cents(income_a)
    income_b_cents = to_cents(income_b)

    combined_income = income_a_cents + income_b_cents

    # Check that combined income is valid.
    if combined_income == 0:
        raise ValueError("Combined income must be greater than zero.")

    # Loop through each expense in the dictionary.
    for expense_name, amount in expenses.items():

        # Validate the expense name.
        name = validate_name(expense_name)

        # Convert the expense into integer cents.
        expense_cents = to_cents(amount)

        # Check for duplicate expense names.
        if name in allocations:
            raise ValueError("Duplicate expense name.")

        # Calculate Person A's contribution.
        # // divides and keeps the whole-number result.
        # Adding half the denominator rounds half-cent cases up.
        person_a_cents = (
            2 * expense_cents * income_a_cents + combined_income
        ) // (2 * combined_income)

        # Person B receives the remaining cents.
        person_b_cents = expense_cents - person_a_cents

        # Store the allocation in a nested dictionary.
        allocations[name] = {
            "expense_cents": expense_cents,
            "person_a_cents": person_a_cents,
            "person_b_cents": person_b_cents
        }

    return allocations
```

```python
def calculate_totals(allocations):
    # Set initial totals to zero.
    total_expenses = 0
    total_a = 0
    total_b = 0

    # Loop through each allocation.
    for allocation in allocations.values():

        # Add the original expense.
        total_expenses += allocation["expense_cents"]

        # Add each person's contribution.
        total_a += allocation["person_a_cents"]
        total_b += allocation["person_b_cents"]

    # Check that all contributions match the expenses.
    if total_a + total_b == total_expenses:
        return total_expenses, total_a, total_b
    else:
        raise ValueError("Allocation totals do not match expenses.")
```

```python
def display_results(name_a, name_b, income_a, income_b, expenses):
    # Validate both names.
    name_a = validate_name(name_a)
    name_b = validate_name(name_b)

    # Calculate income proportions.
    proportion_a, proportion_b = calculate_proportions(
        income_a, income_b
    )

    # Allocate household expenses.
    allocations = allocate_expenses(
        expenses, income_a, income_b
    )

    # Calculate monthly totals.
    total_expenses, total_a, total_b = calculate_totals(
        allocations
    )

    # Calculate combined monthly income.
    combined_income = to_cents(income_a) + to_cents(income_b)

    # Display the application heading.
    print("HOUSEHOLD FAIR-SHARE BUDGET ASSISTANT")
    print("-" * 45)

    # Display each person's income proportion.
    print(f"{name_a}: {proportion_a * 100:.2f}%")
    print(f"{name_b}: {proportion_b * 100:.2f}%")

    print("\nExpense Breakdown:")

    # Display contributions for each shared expense.
    for expense_name, allocation in allocations.items():

        amount_a = allocation["person_a_cents"] / 100
        amount_b = allocation["person_b_cents"] / 100

        print(f"\n{expense_name}")
        print(f"  {name_a}: AUD {amount_a:.2f}")
        print(f"  {name_b}: AUD {amount_b:.2f}")

    # Display the monthly summary.
    print("\nMonthly Summary:")
    print(f"Combined income: AUD {combined_income / 100:.2f}")
    print(f"Total expenses: AUD {total_expenses / 100:.2f}")
    print(f"{name_a}'s contribution: AUD {total_a / 100:.2f}")
    print(f"{name_b}'s contribution: AUD {total_b / 100:.2f}")

    # Return the results for future use by the interface and AI.
    return {
        "name_a": name_a,
        "name_b": name_b,
        "proportion_a": proportion_a,
        "proportion_b": proportion_b,
        "allocations": allocations,
        "combined_income_cents": combined_income,
        "total_expenses_cents": total_expenses,
        "total_a_cents": total_a,
        "total_b_cents": total_b
    }
```
